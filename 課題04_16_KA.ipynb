# **刹那の見切り**

## **概要**

このプログラムは、ゲーム 星のカービィの中で登場するミニゲーム「刹那の見切り」をたいへん"オマージュ"したゲームを作ったものです。

このゲームでは人間の反射神経を競うゲームシステムになっていて、画面に「！！」が表示された時にいかに早くEnterを押せるかの勝負というとても単純なゲームです。

ゲームについての詳しい説明を加えると、シーンは大きく3つに分かれ、タイトル画面と設定画面とゲーム画面を用意しています。それぞれタイトル画面からは設定とゲームの両方にアクセスでき、設定画面とゲーム画面は終了時にタイトル画面に戻るようになっています。

3つのシーンは各関数でまとめられていて、上からtitle()、setting()、game()のなかにコードが記述されています。


## **操作方法**

下のセルを実行すればプレイできます。

設定画面ではゲームの難易度を選択できます。難易度はAをおすすめします

inputで入力する欄が現れた場合は基本的にEnterで次へ進めます。
AやらBを押さなければいけない場合もありますが、画面上に表示される指示に従ってください。

ゲームを開始した後は、画面上に「！！」が現れたのを確認できたなら、すぐにEnterを押して構いません(「！！」と同時にinputの入力欄が出るので)

```python
%reset -f

# 変数とかの準備
from IPython.display import clear_output
import time
import random
LimitTime = 0.6
Difficulty = "A"

# タイトル画面
def title():
  clear_output(wait=True)
  print("\n\nーーー刹那の見切りーーー\n\n\n\nAでスタート　　Bで設定\n\n")
  choose = input()
  if choose=="A" :
    game();
  elif choose=="B" :
    setting();
  else :
    print("タイプミスかも？")
    time.sleep(0.7)
    title();

# 設定
def setting():
  global LimitTime, Difficulty
  clear_output(wait=True)
  print(f'デフォルトは難易度【中】に設定されています　　　現在のモード→{Difficulty}　　　　設定を終えたなら Q を入力\n\n難易度【高】は S を入力(猶予0.45秒)\n難易度【中】は A を入力(猶予0.6秒)\n難易度【低】は B を入力(猶予0.8秒)')
  choose = input()
  if choose=="S" :
    LimitTime=0.45;
    Difficulty="S"
    setting()
  elif choose=="A" :
    LimitTime=0.6;
    Difficulty="A"
    setting()
  elif choose=="B" :
    LimitTime=0.8;
    Difficulty="B"
    setting()
  elif choose=="Q" :
    title();
  else :
    print("タイプミスかも？")
    time.sleep(0.7)
    setting();

# ゲーム画面
def game():
  clear_output(wait=True)
  print("\n\n\n\n　　　　　◎　　　　　　　　　　　　　　　×　　　　　\n_______________________________________________________")
  time.sleep(random.uniform(2, 8))
  clear_output(wait=True)
  TimerStart = time.perf_counter()
  print("\n\n　　　　　ーーーー//　　　!!　　　//ーーーー\n\n　　　　　◎　　　　　　　　　　　　　　　×　　　　　\n_______________________________________________________")
  input()
  Time = time.perf_counter()-TimerStart
  clear_output(wait=True)
  print("\n\n\n\n　　　　　◎　　　　　　　　　　　　　　　×　　　　　\n_______________________________________________________")
  time.sleep(0.2)
  clear_output(wait=True)
  print("\n\n\n\n\n　　　　　　　　　　　　　　　　　　　　　　　　　　　　　　　　")

  time.sleep(0.5)
  clear_output(wait=True)
  print("\n\n\n\n　　　　×　　　　　　　　　　　　　　　　　◎　　　　　\n_______________________________________________________")
  time.sleep(2)
  clear_output(wait=True)
  if Time<LimitTime :
    print(f'\n\n　　　　　　　　　You Win　({Time:.3f}秒)　　　　　　　\n\n　　　　\U0001F4A5　　　　　　　　　　　　　　　　　◎　　　　　\n_______________________________________________________')
  else :
    print(f'\n\n　　　　　　　　　You Lose　({Time:.3f}秒)　　　　　　　\n\n　　　　×　　　　　　　　　　　　　　　　　\U0001F4A5　　　　　\n_______________________________________________________')
  clear_output(wait=True)
  time.sleep(3)
  input("タイトルに戻る")
  title();

# コード本体
title();
```
